## Normalization in the colors
- if all pixel intensity values equal to 0 then color is black 
- if all pixel intensity values equal to 255 then color is white
- Incase if you use float values any value within the float even 255.0 is normalized to 0 or 1
- So we must use `np.uint8` to avoid auto converstion 

```python
# Black Color 
arr= np.zeros((400,400), dtype=np.uint8)
cv2.imshow('Black', arr)
cv2.waitKey()
cv2.destroyAllWindows()
```

```python
# White 
arr= np.full((400,400), 255, dtype=np.uint8)
cv2.imshow('White', arr)
cv2.waitKey()
cv2.destroyAllWindows()
```

```python
# In between Grey 
arr= np.full((400,400), 127, dtype=np.uint8)
cv2.imshow('grey', arr)
cv2.waitKey()
cv2.destroyAllWindows()

```

### Now play with RGB Colors 
- We all know White Color contain 3 Chaneels , R G B
- Their index are B-0, G-1, R-2


```python
redarr= np.zeros((400,400,3)) # here we should assign channels
redarr[:,:,2]=255
cv2.imshow('RED', redarr)
cv2.waitKey()
cv2.destroyAllWindows()
```

```python
# Greeen
Greenarr=np.zeros((400,400,3))
Greenarr[:,:,1]= 255
cv2.imshow('Green', Greenarr)
cv2.waitKey()
cv2.destroyAllWindows()
```

```python
# Blue 
BLUE=np.zeros((400,400,3))
BLUE[:,:,0]= 255
cv2.imshow('BLUE', BLUE)
cv2.waitKey()
cv2.destroyAllWindows()
```

## Blending 

```python
blend1= 0.5*BLUE+0.5*Greenarr
cv2.imshow('BLEND ', blend1)
cv2.waitKey()
cv2.destroyAllWindows()
```

```python
blendblend= 0.5*BLUE+0.5*redarr
cv2.imshow('BLEND ', blendblend)
cv2.waitKey()
cv2.destroyAllWindows()
```

```python
blend= 0.5*redarr+0.5*Greenarr
cv2.imshow('BLEND ', blend)
cv2.waitKey()
cv2.destroyAllWindows()
```

```python
three_mix= 0.5*redarr+0.5*Greenarr+0.5*BLUE
cv2.imshow('MIX ', three_mix)
cv2.waitKey()
cv2.destroyAllWindows()
```

```python
supermix=0.5*blend1+0.5*blendblend+0.5*blend
cv2.imshow('supermix ', three_mix)
cv2.waitKey()
cv2.destroyAllWindows()
```

## BITWISE OPERATORS 

```python
arr= np.zeros((400,400), dtype=np.uint8)
arr[50:200,50:200]=255
cv2.imshow('BLACKWINDOW', arr)
cv2.waitKey()
cv2.destroyAllWindows()
```

```python
arr2= np.zeros((400,400), dtype=np.uint8)
arr2[150:300,150:300]=255
cv2.imshow('BLACKWINDOW', arr2)
cv2.waitKey()
cv2.destroyAllWindows()
```

```python
#Bitwise AND
OPR= cv2.bitwise_and(arr,arr2)
cv2.imshow('bitwise_and', OPR)
cv2.waitKey()
cv2.destroyAllWindows()
```

```python
#Bitwise OR
OPR= cv2.bitwise_or(arr,arr2)
cv2.imshow('bitwise_or', OPR)
cv2.waitKey()
cv2.destroyAllWindows()
```

```python
#Bitwise XOR
OPR= cv2.bitwise_xor(arr,arr2)
cv2.imshow('bitwise_xor', OPR)
cv2.waitKey()
cv2.destroyAllWindows()
```

```python
#Bitwise NOT
OPR= cv2.bitwise_not(arr,arr2)
cv2.imshow('bitwise_not', OPR)
cv2.waitKey()
cv2.destroyAllWindows()
```

```python
#Bitwise NOT
OPR= cv2.bitwise_not(arr2,arr)
cv2.imshow('bitwise_not', OPR)
cv2.waitKey()
cv2.destroyAllWindows()
```
