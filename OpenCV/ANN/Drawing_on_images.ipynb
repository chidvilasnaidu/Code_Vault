### Circle 

```python
canv3= np.full((400,400), 255, dtype=np.uint8)
X=cv2.circle(canv3, (200,200), 78, color=(1,1,1),thickness =5 )
cv2.imshow('Circle ',X)
cv2.waitKey()
cv2.destroyAllWindows()
```

```python
canv3= np.full((400,400), 255, dtype=np.uint8) # while giving colors on mannual Canvas or araays defining mannual channels ( 3D array )is very imp 
X=cv2.circle(canv3, (200,200), 78, color=(1,1,1),thickness =-1 ) # Here we doesnot give channels on mannual canvas , so by default it is grey style image 
cv2.imshow('Circle ',X)
cv2.waitKey()
cv2.destroyAllWindows()
```

```python
canv3= np.full((400,400,3), 255, dtype=np.uint8 )# here we gave channel count so this will give coloured drawings  
X=cv2.circle(canv3, (200,200), 78, color=(255, 0, 0),thickness =5 )
cv2.imshow('Circle ',X)
cv2.waitKey()
cv2.destroyAllWindows()
```

```python
canv3= np.full((400,400,3), 255, dtype=np.uint8 )# here we gave channel count so this will give coloured drawings  
X=cv2.circle(canv3, (100,100), 5, color=(255, 0, 0),thickness =5 )
X=cv2.circle(canv3, (200,100), 5, color=(0, 0, 255),thickness =5 )

cv2.imshow('Circle ',X)
cv2.waitKey()
cv2.destroyAllWindows()
```

### Lines

```python
# Single line 
canv4= np.full((400,400,3), 255, dtype=np.uint8 )
X=cv2.line(canv4, (50,50), (300,300), color=(0,255,0),thickness =3 )
cv2.imshow('Line ',X)
cv2.waitKey()
cv2.destroyAllWindows()
```

```python
# Multi LINES 
canv4= np.full((400,400,3), 255, dtype=np.uint8 )
X=cv2.line(canv4, (50,50), (300,300), color=(0,0,255),thickness =3 )
Y=cv2.line(canv4, (50,300),(300,50) , color=(0,0,255),thickness =3 )
cv2.imshow('Cross ', Y)
cv2.waitKey()
cv2.destroyAllWindows()

```

### Text

```python
canv4= np.full((1200,1200,3), 0, dtype=np.uint8 )
txt=cv2.putText(canv4, "BOUNJUR",org=(200,300),fontFace=cv2.FONT_HERSHEY_SCRIPT_COMPLEX, fontScale =5, color =(70,155,30))
cv2.imshow('BOUNJUR ', txt)
cv2.waitKey()
cv2.destroyAllWindows()
```
