### Affiliate Transformation 

### 1. Translation
    - Here We will transform the place of an image from one point to another 

```python
import cv2
import numpy as np



```

```python
img= cv2.imread(r"C:\Users\chidv\Downloads\OpenCV_Logo.png")
cv2.imshow("LOGO", img)
cv2.waitKey()
cv2.destroyAllWindows()
```

### M tells OpenCV how much the x and y coordinates of each pixel should be shifted.

```python
m=np.array([[0,1,50],[1,0,100]], dtype=np.float32)
m
```

```python
X=cv2.warpAffine(img, m, (300,300))
cv2.imshow("translated image", X)
cv2.waitKey()
cv2.destroyAllWindows()
```

### 4. Shear
- so for shear, there’s no separate cv2.shear() function. You use a 2 by 3 transformation matrix M and pass that to cv2.warpAffine(). You basically choose the numbers based on how much effect you want. For example, for horizontal shear, you might use something like M equals np.float32 of 1, 0.3, 0 in the first row, and 0, 1, 0 in the second row

```python
M= np.array(([[ 1,0.3,0],[0,1,0]]),dtype=np.float32) 
M
```

```python
X=cv2.warpAffine(img, M, (660,660))
cv2.imshow("Shear",X)
cv2.waitKey()
cv2.destroyAllWindows()
```
