# Canny:

- Is a complete edge-detection algorithm.
- It tries to produce thin, clean edges.
- It reduces noise before detecting edges.
- It uses gradient calculation, non-maximum suppression, and thresholding.
- It detects edges in different directions, not just horizontal/vertical.

### Thresholds in Canny are used to decide which detected intensity changes are strong enough to be treated as edges.
- < 100 → rejected
- 100–200 → weak edge
- />200 → strong edge

#### 1. Theshold Defines the minimum strength for an edge to be considered.
Below this value → usually not considered an edge.
#### 2. Upper threshold (threshold2)
Defines the strength required for an edge to be considered a strong edge

```python
img=cv2.imread(r"C:\Users\chidv\Batch405\CV\ANN\Edge.jpg")
canny= cv2.Canny(img, 100,200)
cv2.imshow("canny ", canny)
cv2.waitKey()
cv2.destroyAllWindows()
```

## Blur :  is commonly applied before edge detection to reduce noise and prevent false edges

```python
# Average Blur 
img=cv2.imread(r"C:\Users\chidv\Batch405\CV\ANN\puppy.jpg")
img = cv2.resize(img, None, fx=0.5, fy=0.5)
AVGBLR= cv2.blur(img,(5,5))
cv2.imshow("AVGBLUR ", AVGBLR)
cv2.waitKey()
cv2.destroyAllWindows()
```

```python
# Gaussian Blur

img=cv2.imread(r"C:\Users\chidv\Batch405\CV\ANN\puppy.jpg")
img = cv2.resize(img, None, fx=0.5, fy=0.5)
Gaussian= cv2.GaussianBlur(img,(5,5),4) #(5,5 blur facctor 
cv2.imshow("Gaussian  ", Gaussian)
cv2.waitKey()
cv2.destroyAllWindows()
```
