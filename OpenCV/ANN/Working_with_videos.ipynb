## Applying Canny Edge Detection on Videos

```python
cap=cv2.VideoCapture(r"C:\Users\chidv\Batch405\CV\ANN\flower.mp4")
while True:
    istrue, frames=cap.read()
    if istrue:
        resize= cv2.resize(frames, (450,600), fx=0.5,fy=0.5)
        canny= cv2.Canny(resize,100,200)
        cv2.imshow('Flower Detection', canny)
        if cv2.waitKey(40) & 0xFF == ord('q'):
            break
    else:
        break
cv2.destroyAllWindows()
```
