```python
# 1. Import required libraries
import cv2
from ultralytics import YOLO
from google.colab.patches import cv2_imshow

# 3. Specify the video path
video_path = "/content/Helmet_bike.mp4"

# 4. Open the video
cap = cv2.VideoCapture(video_path)

# 5. Initialize frame counter
count = 0

# 6. Read and process video frames
while True:

    # Read the next frame
    ret, frame = cap.read()

    # Stop if video ends or frame cannot be read
    if not ret:
        break

    # Increment frame counter
    count += 1

    # Process every 50th frame
    if count % 50 == 0:

        # Perform object detection using YOLO
        result = model(frame)

        # Draw bounding boxes and labels
        annotated_frame = result[0].plot()

        # Display detected frame
        cv2_imshow(annotated_frame)

    # Press 'q' to stop video
    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

# 7. Release video resources
cap.release()

# 8. Close OpenCV windows
cv2.destroyAllWindows()
```
